```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# loading training data
train = pd.read_csv(
    r"H:\codesoft\Genre Classification Dataset\train_data.txt",
    sep=" ::: ",
    engine="python",
    names=["id","genre","plot"]
)

# taking plots and genres
X = train["plot"]
y = train["genre"]

# converting text to numbers
tfidf = TfidfVectorizer(stop_words="english")
X_vec = tfidf.fit_transform(X)

# training the model
model = LogisticRegression(max_iter=1000)
model.fit(X_vec, y)



# loading test data
test = pd.read_csv(
    r"H:\codesoft\Genre Classification Dataset\test_data.txt",
    sep=" ::: ",
    engine="python",
    names=["id","plot"]
)

solution = pd.read_csv(
    r"H:\codesoft\Genre Classification Dataset\test_data_solution.txt",
    sep=" ::: ",
    engine="python",
    names=["id","genre"]
)

# converting test plots
X_test = tfidf.transform(test["plot"])

# predicting genres
pred = model.predict(X_test)

# checking accuracy
acc = accuracy_score(solution["genre"], pred)
print("accuracy:", acc)

# letting user test the model
plot = input("\nEnter a movie plot: ")

plot_vec = tfidf.transform([plot])
result = model.predict(plot_vec)

print("Predicted genre:", result[0])
```
